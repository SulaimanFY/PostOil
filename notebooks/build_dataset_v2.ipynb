Pivot databank set

```python
# World Bank Databank export: wide years → one row per (country, year)

id_cols = ["Country Name", "Country Code", "Series Name", "Series Code"]
year_cols = [c for c in databank.columns if c not in id_cols]

databank_long = databank.melt(
    id_vars=id_cols,
    value_vars=year_cols,
    var_name="year_raw",
    value_name="value",
)

# "1960 [YR1960]" → 1960
databank_long["year"] = (
    databank_long["year_raw"].astype(str).str.extract(r"(\d{4})")[0].astype("Int64")
)

databank_long["value"] = pd.to_numeric(
    databank_long["value"].replace(["..", ""], pd.NA),
    errors="coerce",
)

series_code_to_col = {
    "NY.GDP.PETR.RT.ZS": "oil_rents_gdp",
    "TX.VAL.FUEL.ZS.UN": "fuel_exports_pct",
    "NV.SRV.TOTL.ZS": "services_gdp",
    "NV.IND.MANF.ZS": "manufacturing_gdp",
    "NE.EXP.GNFS.ZS": "exports_gdp",
    "NE.IMP.GNFS.ZS": "imports_gdp",
    "NE.TRD.GNFS.ZS": "trade_openness_gdp",
    "NE.EXP.GNFS.CD": "exports_usd",
    "NE.IMP.GNFS.CD": "imports_usd",
    "NY.GDP.MKTP.CD": "gdp",
    "NY.GDP.PCAP.CD": "gdp_per_capita",
}

wdi = (
    databank_long[databank_long["Series Code"].isin(series_code_to_col)]
    .assign(variable=lambda d: d["Series Code"].map(series_code_to_col))
    .pivot_table(
        index=["Country Name", "year"],
        columns="variable",
        values="value",
        aggfunc="first",
    )
    .reset_index()
    .rename(columns={"Country Name": "country"})
)

# The pivot above puts the mapped codes into columns across the variables axis;
# There is no longer a "variable" column to drop. Remove that line and just sort:
wdi = wdi.sort_values(["country", "year"])

wdi.head()
```

Renaming columns (especially key columns) before merge

```python
# UNCTAD → merge keys + export concentration
exports_concentration = exports_concentration.rename(
    columns={
        "Economy_Label": "country",
        "Year": "year",
        "Concentration_Index_Value": "export_concentration",
    }
)

# WIPO GII extract
innovation = innovation.rename(
    columns={
        "innovation_score": "innovation_index",
    }
)

# Oxford AI readiness extract
ai_readiness = ai_readiness.rename(
    columns={
        "total_score": "ai_readiness",
        "rank": "ai_readiness_rank",
    }
)
```

What column to keep?

```python
wdi_cols = [
    "country",
    "year",
    "oil_rents_gdp",
    "fuel_exports_pct",
    "services_gdp",
    "manufacturing_gdp",
    "exports_gdp",
    "imports_gdp",
    "trade_openness_gdp",
    "exports_usd",
    "imports_usd",
    "gdp",
    "gdp_per_capita",
]
wdi = wdi[wdi_cols]

exports_cols = ["country", "year", "export_concentration"]
exports_concentration = exports_concentration[exports_cols]

innovation_cols = ["country", "year", "innovation_index", "innovation_rank"]
innovation = innovation[innovation_cols]

ai_readiness_cols = [
    "country",
    "year",
    "ai_readiness",
    "ai_readiness_rank",
    "government_pillar",
    "technology_sector_pillar",
    "data_infrastructure_pillar",
    "policy_capacity",
    "ai_infrastructure",
    "governance",
    "public_sector_adoption",
    "development_diffusion",
    "resilience",
]
ai_readiness = ai_readiness[ai_readiness_cols]
```

```python
for name, _df in [
    ("wdi", wdi),
    ("exports_concentration", exports_concentration),
    ("innovation", innovation),
    ("ai_readiness", ai_readiness),
]:
    assert not _df.duplicated(["country", "year"]).any(), f"duplicate keys in {name}"
```

```python
merged = (
    wdi.merge(exports_concentration, on=["country", "year"], how="outer")
    .merge(innovation, on=["country", "year"], how="outer")
    .merge(ai_readiness, on=["country", "year"], how="outer")
)
merged = merged.sort_values(["country", "year"], kind="mergesort").reset_index(drop=True)
merged.head()
```

SQLite — four source tables (`datasets/postoil.db`)

```python
import sqlite3

db_path = "../datasets/postoil.db"

with sqlite3.connect(db_path) as conn:
    wdi.to_sql("wdi_core", conn, if_exists="replace", index=False)
    exports_concentration.to_sql("export_concentration", conn, if_exists="replace", index=False)
    ai_readiness.to_sql("ai_readiness", conn, if_exists="replace", index=False)
    innovation.to_sql("innovation", conn, if_exists="replace", index=False)

    for tbl in ("wdi_core", "export_concentration", "ai_readiness", "innovation"):
        conn.execute(
            f"CREATE INDEX IF NOT EXISTS idx_{tbl}_country_year ON {tbl} (country, year)"
        )

db_path
```
